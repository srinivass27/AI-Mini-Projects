```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "j-hartmann/emotion-english-distilroberta-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

print("Emotion detection model loaded successfully!")
```

```python
def detect_emotion(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=1)

    predicted_class = torch.argmax(probabilities, dim=1).item()
    confidence = probabilities[0][predicted_class].item()

    emotion = model.config.id2label[predicted_class]

    return emotion, confidence
```

```python
text = "I finally got the job! I am so happy!"

emotion, confidence = detect_emotion(text)

print("Text:")
print(text)

print("\nDetected Emotion:")
print(emotion)

print(f"Confidence: {confidence:.2%}")
```

```python
test_texts = [
    "I am extremely angry about what happened.",
    "I am scared about what might happen tomorrow.",
    "I feel really sad and lonely today.",
    "Wow! I did not expect that at all!",
    "This food smells absolutely disgusting."
]

for text in test_texts:
    emotion, confidence = detect_emotion(text)
    print(f"{text}")
    print(f"→ {emotion} ({confidence:.2%})")
    print("-" * 60)
```

```python
while True:
    text = input("\nEnter a sentence (or type 'exit' to stop): ")

    if text.lower() == "exit":
        print("Emotion detector stopped.")
        break

    emotion, confidence = detect_emotion(text)

    print("\nDetected Emotion:", emotion)
    print(f"Confidence: {confidence:.2%}")
```
