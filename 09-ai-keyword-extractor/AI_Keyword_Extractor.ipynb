```python
from keybert import KeyBERT

keyword_model = KeyBERT(model="all-MiniLM-L6-v2")

print("Keyword extraction model loaded successfully!")
```

```python
def extract_keywords(text, top_n=5):
    keywords = keyword_model.extract_keywords(
        text,
        keyphrase_ngram_range=(1, 3),
        stop_words="english",
        top_n=top_n
    )

    return keywords
```

```python
def extract_keywords(text, top_n=5):
    keywords = keyword_model.extract_keywords(
        text,
        keyphrase_ngram_range=(1, 3),
        stop_words="english",
        top_n=top_n
    )

    return keywords
```

```python
keyphrase_ngram_range=(1, 3)
```

```python
stop_words="english"
```

```python
top_n=5
```

```python
text = """
Artificial intelligence and machine learning are transforming healthcare
through automated diagnosis and medical image analysis. Deep learning
models can help doctors detect diseases from chest X-ray images.
"""

keywords = extract_keywords(text, top_n=5)

print("Extracted Keywords:\n")

for keyword, score in keywords:
    print(f"{keyword} → {score:.2f}")
```

```python
while True:
    text = input("\nEnter text (or type 'exit' to stop): ")

    if text.lower() == "exit":
        print("Keyword extractor stopped.")
        break

    keywords = extract_keywords(text, top_n=5)

    print("\nTop Keywords:")

    for keyword, score in keywords:
        print(f"- {keyword} ({score:.2f})")
```

```python
def display_keywords(text, top_n=5):
    keywords = extract_keywords(text, top_n=top_n)

    print("Top Keywords:")
    print("-" * 30)

    for keyword, score in keywords:
        print(f"{keyword:<30} {score:.2f}")
```

```python
text = """
Artificial intelligence, machine learning, and data science are
transforming healthcare through automated diagnosis and medical
image analysis.
"""

display_keywords(text, top_n=5)
```
