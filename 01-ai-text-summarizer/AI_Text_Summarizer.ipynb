```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Summarizer loaded successfully!")
```

```python
def summarize_text(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        max_length=1024,
        truncation=True
    )

    summary_ids = model.generate(
        inputs["input_ids"],
        max_length=150,
        min_length=30,
        num_beams=4,
        early_stopping=True
    )

    summary = tokenizer.decode(
        summary_ids[0],
        skip_special_tokens=True
    )

    return summary
```

```python
text = """
Artificial intelligence is rapidly changing many industries around the world.
Companies are using AI to automate repetitive tasks, analyze large amounts of
data, improve customer service, detect fraud, and assist employees in making
better decisions. Recent advances in large language models have also made it
possible to build applications that can understand and generate human language.
However, organizations still need to consider issues such as privacy, security,
accuracy, bias, and the responsible use of AI.
"""

summary = summarize_text(text)

print("ORIGINAL TEXT:\n")
print(text)

print("\n" + "=" * 60)

print("\nAI-GENERATED SUMMARY:\n")
print(summary)
```

```python
while True:
    print("\n" + "=" * 60)
    print("AI TEXT SUMMARIZER")
    print("=" * 60)

    text = input("\nEnter the text you want to summarize:\n\n")

    if text.lower() == "exit":
        print("\nExiting summarizer. Goodbye!")
        break

    if len(text.strip()) < 50:
        print("\nPlease enter a longer piece of text.")
        continue

    summary = summarize_text(text)

    print("\n" + "-" * 60)
    print("AI-GENERATED SUMMARY:")
    print("-" * 60)
    print(summary)
```

```python
def run_summarizer():
    print("=" * 60)
    print("🤖 AI TEXT SUMMARIZER")
    print("=" * 60)
    print("Enter your text below.")
    print("Type 'exit' to quit.")

    while True:
        print("\n" + "-" * 60)
        text = input("Enter text: ")

        if text.lower().strip() == "exit":
            print("\nThank you for using AI Text Summarizer!")
            break

        if len(text.strip()) < 50:
            print("⚠️ Please enter at least 50 characters.")
            continue

        print("\n⏳ Generating summary...")

        summary = summarize_text(text)

        print("\n✅ SUMMARY")
        print("-" * 60)
        print(summary)
```

```python
run_summarizer()
```

A simple AI-powered text summarization application built using Python,
Hugging Face Transformers, and the BART-large-CNN pretrained model.

## Features

- Summarizes long text automatically
- Interactive user input
- Uses a pretrained Transformer model
- No external API required
- Runs directly in Google Colab

## Technologies

- Python
- PyTorch
- Hugging Face Transformers
- BART-large-CNN
- Google Colab
