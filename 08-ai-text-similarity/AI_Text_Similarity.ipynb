```python
from sentence_transformers import SentenceTransformer

model_name = "all-MiniLM-L6-v2"

model = SentenceTransformer(model_name)

print("Text similarity model loaded successfully!")
```

```python
from sentence_transformers import util

def calculate_similarity(text1, text2):
    embedding1 = model.encode(text1, convert_to_tensor=True)
    embedding2 = model.encode(text2, convert_to_tensor=True)

    similarity = util.cos_sim(embedding1, embedding2)

    return similarity.item()
```

```python
text1 = "I love playing basketball."
text2 = "Basketball is my favorite sport."

similarity = calculate_similarity(text1, text2)

print("Text 1:", text1)
print("Text 2:", text2)
print(f"\nSimilarity: {similarity:.2%}")
```

```python
text1 = "I love playing basketball."
text2 = "The computer needs a software update."

similarity = calculate_similarity(text1, text2)

print("Text 1:", text1)
print("Text 2:", text2)
print(f"\nSimilarity: {similarity:.2%}")
```

```python
while True:
    text1 = input("\nEnter the first sentence (or type 'exit' to stop): ")

    if text1.lower() == "exit":
        print("Similarity checker stopped.")
        break

    text2 = input("Enter the second sentence: ")

    similarity = calculate_similarity(text1, text2)

    print(f"\nSimilarity Score: {similarity:.2%}")
```
