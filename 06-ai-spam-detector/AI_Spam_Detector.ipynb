```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "mshenoda/roberta-spam"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

print("Spam detection model loaded successfully!")
```

```python
def detect_spam(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=1)

    predicted_class = torch.argmax(probabilities, dim=1).item()
    confidence = probabilities[0][predicted_class].item()

    label = model.config.id2label[predicted_class]

    return label, confidence
```

```python
message = "Congratulations! You have won a free prize. Click the link now!"

label, confidence = detect_spam(message)

print("Message:")
print(message)

print("\nPrediction:")
print(label)

print(f"Confidence: {confidence:.2%}")
```

```python
def detect_spam(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=1)

    predicted_class = torch.argmax(probabilities, dim=1).item()
    confidence = probabilities[0][predicted_class].item()

    if predicted_class == 1:
        label = "SPAM"
    else:
        label = "NOT SPAM"

    return label, confidence
```

```python
message = "Congratulations! You have won a free prize. Click the link now!"

label, confidence = detect_spam(message)

print("Message:")
print(message)

print("\nPrediction:")
print(label)

print(f"Confidence: {confidence:.2%}")
```

```python
while True:
    message = input("\nEnter a message (or type 'exit' to stop): ")

    if message.lower() == "exit":
        print("Spam detector stopped.")
        break

    label, confidence = detect_spam(message)

    print("\nPrediction:", label)
    print(f"Confidence: {confidence:.2%}")
```
